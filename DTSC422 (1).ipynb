```python
import shutil

path = "train_benchmark.py"
shutil.copy(path, path + ".bak2")   # backup

src = open(path).read()

# Edit 1: compute warmup_steps before TrainingArguments
old1 = "    training_args = TrainingArguments("
new1 = (
    "    total_steps = steps_per_epoch * args.num_train_epochs\n"
    "    warmup_steps = max(1, int(total_steps * args.warmup_ratio)) if args.warmup_ratio > 0 else 0\n"
    "    training_args = TrainingArguments("
)
# Edit 2: warmup_ratio -> warmup_steps inside TrainingArguments
old2 = "warmup_ratio=args.warmup_ratio,"
new2 = "warmup_steps=warmup_steps,"
# Edit 3: Trainer(tokenizer=...) -> processing_class=...
old3 = "tokenizer=tokenizer,"
new3 = "processing_class=tokenizer,"

for name, old, new in [("1 warmup calc", old1, new1), ("2 warmup arg", old2, new2), ("3 trainer tokenizer", old3, new3)]:
    n = src.count(old)
    print(f"Edit {name}: found {n} time(s)")
    if n != 1:
        raise SystemExit(f"Edit {name} expected exactly 1 match, so nothing was written.")
    src = src.replace(old, new)

open(path, "w").write(src)
print("Patched.")
```

```python
import pandas as pd, torch, numpy as np
from transformers import AutoTokenizer, AutoModelForSequenceClassification

ck = "indicbertv2-odia-nli-failed/checkpoint-18405"
tok = AutoTokenizer.from_pretrained(ck)
model = AutoModelForSequenceClassification.from_pretrained(ck).eval()

df = pd.read_parquet("indicxnli-test.parquet")
print(df.columns.tolist(), len(df))
df = df.sample(1000, random_state=0)

preds = []
with torch.no_grad():
    for i in range(0, len(df), 32):
        b = df.iloc[i:i+32]
        enc = tok(b["premise"].tolist(), b["hypothesis"].tolist(),
                  truncation=True, max_length=128, padding=True, return_tensors="pt")
        preds += model(**enc).logits.argmax(-1).tolist()
print("accuracy:", (np.array(preds) == df["label"].to_numpy()).mean())
print(np.bincount(preds, minlength=3))
```

```python
import json
from sklearn.metrics import classification_report

df = pd.read_parquet("indicxnli-test.parquet")   # all 5,010 rows
preds = []
with torch.no_grad():
    for i in range(0, len(df), 32):
        b = df.iloc[i:i+32]
        enc = tok(b["premise"].tolist(), b["hypothesis"].tolist(),
                  truncation=True, max_length=128, padding=True, return_tensors="pt")
        preds += model(**enc).logits.argmax(-1).tolist()

y = df["label"].to_numpy()
acc = (np.array(preds) == y).mean() * 100
print(f"Full test accuracy: {acc:.2f}%")
print(classification_report(y, preds, digits=4))

json.dump({"benchmark_test_accuracy_pct": acc, "checkpoint": ck, "n_test": len(df)},
          open("benchmark_ckpt18405.json", "w"), indent=2)
```

```python
import json, pandas as pd, numpy as np, torch

# 1) benchmark predictions (only valid if df/preds are still the full 5,010-row run)
assert len(df) == 5010 and len(preds) == 5010
out = df[["premise", "hypothesis", "label"]].copy()
out["predicted_label"] = preds
out.to_csv("predictions_benchmark_ckpt18405.csv", index=False)

# 2) native predictions
nat = pd.read_json("03_nli_groups_3_9_updated.jsonl", lines=True)
lm = {"entailment": 0, "neutral": 1, "contradiction": 2}
nat["gold"] = nat["label"].map(lm)
nat_preds = []
with torch.no_grad():
    for i in range(0, len(nat), 32):
        b = nat.iloc[i:i+32]
        enc = tok(b["premise"].tolist(), b["hypothesis"].tolist(),
                  truncation=True, max_length=128, padding=True, return_tensors="pt")
        nat_preds += model(**enc).logits.argmax(-1).tolist()
nat["predicted_label"] = nat_preds
nat.to_csv("predictions_native_ckpt18405.csv", index=False)
print("native accuracy:", (nat["gold"] == nat["predicted_label"]).mean())

# 3) readable experiment configuration
a = torch.load(f"{ck}/training_args.bin", weights_only=False)
json.dump(a.to_dict(), open("experiment_config_ckpt18405.json", "w"), indent=2, default=str)
```
